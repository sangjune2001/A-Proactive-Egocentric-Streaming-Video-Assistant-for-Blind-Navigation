```python
# 클래스

VEHICLE_CLASSES = {
    1: "bicycle",
    2: "car",
    3: "motorcycle",
    5: "bus",
    7: "truck",
}
OBSTACLE_CLASSES = {
    13: "bench",
    58: "potted plant",
    56: "chair",
}

TRAFFIC_LIGHT_CLASS = 9

TARGET_CLASSES = (
    list(VEHICLE_CLASSES)
    + list(OBSTACLE_CLASSES)
    + [TRAFFIC_LIGHT_CLASS]
)

# 검출 설정

CONF_THRESHOLD = 0.5 # confidence가 0.4 이하면 제외.
IMAGE_SIZE = 640 # 원본 영상 해상도를 그대로 사용하지 않고, 해당 크기로 리사이즈. 640이 적

# 기존 접근 트리거 기준
GROWTH_THRESHOLD = 1.10  #Bbox 면적이 1.2배(=20%) 이상 증가하면 접근 조건 만족
TRIGGER_STREAK = 5 # 접근 조건 3회 연속 만족시 Trigger 발생

print("신호등 포함 검출 클래스:", [model.names[c] for c in TARGET_CLASSES])
```

```python
# 검출된 신호등 Bbox 안에서 빨간색,초록색 픽셀 수를 비교하여 현재 색상을 판단

def classify_light_color(frame, box):
    x1, y1, x2, y2 = map(int, box)
    height, width = frame.shape[:2]

    x1 = max(0, min(x1, width - 1))
    x2 = max(0, min(x2, width))
    y1 = max(0, min(y1, height - 1))
    y2 = max(0, min(y2, height))

    # 신호등 ROI
    roi = frame[y1:y2, x1:x2]

    # 신호등 색상 HSV로 변환 후 특정 기준에 따라 Red인지 Green인지 return
    hsv = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)
    h, s, v = cv2.split(hsv)

    red_mask = (
        ((h < 10) | (h > 170))
        & (s > 70)
        & (v > 100)
    )

    green_mask = (
        (h > 35) & (h < 90)
        & (s > 50)
        & (v > 80)
    )

    red_score = red_mask.sum()
    green_score = green_mask.sum()

    min_pixels = max(
        3,
        int(roi.shape[0] * roi.shape[1] * 0.005)
    )

    if max(red_score, green_score) < min_pixels:
        return "unknown"

    if red_score > green_score * 1.2:
        return "red"

    if green_score > red_score * 1.2:
        return "green"

    return "unknown"
```

```python
import cv2
import numpy as np

# 접근 판단 설정
APPROACH_WINDOW = 15
APPROACH_AVG = 5
CENTER_SPEED_MAX = 0.03
MAX_OBSERVATION_GAP = 0.3

# 배경 움직임 추정 설정: 초기 실험값
GMC_IMAGE_SIZE = 640
GMC_MIN_POINTS = 20
GMC_MIN_INLIER_RATIO = 0.6
GMC_MAX_FB_ERROR = 1.5

assert APPROACH_WINDOW >= 2 * APPROACH_AVG


class BackgroundMotion:
    def __init__(self):
        self.prev_gray = None
        self.prev_mask = None

    def update(self, frame, boxes):
        """이전 프레임 -> 현재 프레임의 배경 변환 추정."""
        h, w = frame.shape[:2]
        scale = min(1.0, GMC_IMAGE_SIZE / max(h, w))
        sw = max(2, round(w * scale))
        sh = max(2, round(h * scale))

        gray = cv2.cvtColor(
            cv2.resize(frame, (sw, sh)),
            cv2.COLOR_BGR2GRAY,
        )

        # 검출된 객체 영역은 배경 특징점 추출에서 제외
        mask = np.full((sh, sw), 255, dtype=np.uint8)
        sx, sy = sw / w, sh / h

        for box in boxes:
            x1, y1, x2, y2 = map(float, box)
            margin = 5
            left = max(0, int(np.floor(x1 * sx)) - margin)
            top = max(0, int(np.floor(y1 * sy)) - margin)
            right = min(sw, int(np.ceil(x2 * sx)) + margin)
            bottom = min(sh, int(np.ceil(y2 * sy)) + margin)

            if right > left and bottom > top:
                mask[top:bottom, left:right] = 0

        prev_gray, prev_mask = self.prev_gray, self.prev_mask
        self.prev_gray, self.prev_mask = gray, mask

        info = {
            "camera_ok": False,
            "camera_reason": "INIT",
            "camera_points": 0,
            "camera_inliers": 0,
            "camera_inlier_ratio": 0.0,
        }

        if prev_gray is None:
            return None, info

        if prev_gray.shape != gray.shape:
            info["camera_reason"] = "SIZE_CHANGED"
            return None, info

        points0 = cv2.goodFeaturesToTrack(
            prev_gray,
            maxCorners=500,
            qualityLevel=0.01,
            minDistance=10,
            mask=prev_mask,
            blockSize=7,
        )

        if points0 is None or len(points0) < GMC_MIN_POINTS:
            info["camera_reason"] = "FEW_FEATURES"
            return None, info

        # 이전 -> 현재
        points1, status1, _ = cv2.calcOpticalFlowPyrLK(
            prev_gray, gray, points0, None,
            winSize=(21, 21),
            maxLevel=3,
        )

        if points1 is None or status1 is None:
            info["camera_reason"] = "FLOW_FAILED"
            return None, info

        valid = (
            status1.ravel().astype(bool)
            & np.isfinite(points1.reshape(-1, 2)).all(axis=1)
        )
        p0 = points0.reshape(-1, 2)[valid]
        p1 = points1.reshape(-1, 2)[valid]

        if len(p0) < GMC_MIN_POINTS:
            info["camera_reason"] = "FEW_TRACKS"
            return None, info

        # 현재 -> 이전 역추적으로 부정확한 특징점 제거
        back, status2, _ = cv2.calcOpticalFlowPyrLK(
            gray, prev_gray,
            p1.reshape(-1, 1, 2), None,
            winSize=(21, 21),
            maxLevel=3,
        )

        if back is None or status2 is None:
            info["camera_reason"] = "BACK_FLOW_FAILED"
            return None, info

        fb_error = np.linalg.norm(back.reshape(-1, 2) - p0, axis=1)
        valid = (
            status2.ravel().astype(bool)
            & np.isfinite(fb_error)
            & (fb_error <= GMC_MAX_FB_ERROR)
            & (p1[:, 0] >= 0)
            & (p1[:, 0] < sw)
            & (p1[:, 1] >= 0)
            & (p1[:, 1] < sh)
        )
        p0, p1 = p0[valid], p1[valid]

        # 현재 프레임에서도 객체 내부로 들어간 특징점 제외
        if len(p1):
            indices = np.floor(p1).astype(int)
            background = mask[indices[:, 1], indices[:, 0]] > 0
            p0, p1 = p0[background], p1[background]

        info["camera_points"] = len(p0)

        if len(p0) < GMC_MIN_POINTS:
            info["camera_reason"] = "FEW_BACKGROUND_TRACKS"
            return None, info

        matrix, inliers = cv2.estimateAffinePartial2D(
            p0, p1,
            method=cv2.RANSAC,
            ransacReprojThreshold=2.0,
            maxIters=2000,
            confidence=0.99,
            refineIters=10,
        )

        if matrix is None or inliers is None:
            info["camera_reason"] = "RANSAC_FAILED"
            return None, info

        inlier_mask = inliers.ravel().astype(bool)
        count = int(inlier_mask.sum())
        ratio = count / len(p0)

        info["camera_inliers"] = count
        info["camera_inlier_ratio"] = ratio

        if (
            not np.isfinite(matrix).all()
            or count < GMC_MIN_POINTS
            or ratio < GMC_MIN_INLIER_RATIO
        ):
            info["camera_reason"] = "LOW_CONFIDENCE"
            return None, info

        # 특징점이 너무 좁은 영역에 몰린 경우 제외
        spread = np.ptp(p0[inlier_mask], axis=0)
        if spread[0] < sw * 0.2 or spread[1] < sh * 0.2:
            info["camera_reason"] = "LOCAL_FEATURES"
            return None, info

        # 프레임 간 비정상적으로 큰 확대/축소 제외
        estimated_scale = float(np.hypot(matrix[0, 0], matrix[1, 0]))
        if not 0.9 <= estimated_scale <= 1.1:
            info["camera_reason"] = "EXTREME_SCALE"
            return None, info

        # 축소 이미지 좌표의 변환을 원본 좌표의 변환으로 환산
        small_transform = np.eye(3)
        small_transform[:2] = matrix

        resize_transform = np.diag([sx, sy, 1.0])
        full_transform = (
            np.linalg.inv(resize_transform)
            @ small_transform
            @ resize_transform
        )

        info["camera_ok"] = True
        info["camera_reason"] = "OK"
        return full_transform[:2], info


def update_approach_coordinates(matrix, camera_info, frame_index, fps):
    """과거 중심 좌표를 매 프레임 현재 화면 좌표로 이동."""
    if not camera_info["camera_ok"]:
        # 잘못된 변환을 누적하지 않도록 기록 초기화
        approach_history.clear()
        vehicle_streak.clear()
        return

    for key in list(approach_history):
        history = approach_history[key]

        if (
            not history
            or (frame_index - history[-1]["frame"]) / fps
            > MAX_OBSERVATION_GAP
        ):
            del approach_history[key]
            vehicle_streak.pop(key, None)
            continue

        for observation in history:
            observation["corrected_center"] = (
                matrix[:, :2] @ observation["corrected_center"]
                + matrix[:, 2]
            )


def check_approach(
    box, class_id, track_id, frame_index, fps, w, h, camera_info
):
    key = track_id
    name = (
        VEHICLE_CLASSES[class_id]
        if class_id in VEHICLE_CLASSES
        else OBSTACLE_CLASSES[class_id]
    )

    x1, y1, x2, y2 = map(float, box)
    area = max(0.0, x2 - x1) * max(0.0, y2 - y1)
    center = np.array([(x1 + x2) / 2, (y1 + y2) / 2])
    time_s = (frame_index - 1) / fps

    previous_frame = approach_last_seen.get(key)
    missing_frames = (
        0 if previous_frame is None
        else max(0, frame_index - previous_frame - 1)
    )
    approach_last_seen[key] = frame_index

    if missing_frames > 0:
        vehicle_streak[key] = 0

    history = approach_history[key]
    history.append({
        "frame": frame_index,
        "time": time_s,
        "area": area,
        "raw_center": center.copy(),
        "corrected_center": center.copy(),
    })

    message = None
    growth = None
    raw_speed = None
    corrected_speed = None
    reasons = []

    if not camera_info["camera_ok"]:
        reasons.append("CAMERA_" + camera_info["camera_reason"])

    if len(history) < APPROACH_WINDOW:
        reasons.append(f"WARMING {len(history)}/{APPROACH_WINDOW}")
    else:
        observations = list(history)
        past = observations[:APPROACH_AVG]
        recent = observations[-APPROACH_AVG:]

        growth = float(
            np.mean([o["area"] for o in recent])
            / max(np.mean([o["area"] for o in past]), 1.0)
        )

        dt = (
            np.mean([o["time"] for o in recent])
            - np.mean([o["time"] for o in past])
        )
        size = np.array([w, h], dtype=float)

        def get_speed(field):
            past_center = np.mean([o[field] for o in past], axis=0)
            recent_center = np.mean([o[field] for o in recent], axis=0)
            return float(
                np.linalg.norm((recent_center - past_center) / size)
                / max(dt, 1e-6)
            )

        raw_speed = get_speed("raw_center")
        corrected_speed = get_speed("corrected_center")

        growing = growth >= GROWTH_THRESHOLD
        stable = corrected_speed <= CENTER_SPEED_MAX
        passes = growing and stable and camera_info["camera_ok"]

        if passes:
            vehicle_streak[key] += 1
        else:
            vehicle_streak[key] = 0

        if not growing:
            reasons.append("GROWTH_LOW")
        if not stable:
            reasons.append("CENTER_SPEED_HIGH")

        if passes:
            reasons.append(
                "CONDITIONS_MET"
                if vehicle_streak[key] >= TRIGGER_STREAK
                else "STREAK_WAIT"
            )

        if (
            vehicle_streak[key] >= TRIGGER_STREAK
            and key not in vehicle_triggered
        ):
            vehicle_triggered.add(key)
            message = (
                f"{name} ID={track_id} approach candidate "
                f"g={growth:.2f} v={corrected_speed:.3f}"
            )

    if missing_frames:
        reasons.append(f"OBSERVATION_GAP={missing_frames}")

    if growth is None:
        label = f"ID={track_id} warming={len(history)}/{APPROACH_WINDOW}"
    else:
        label = (
            f"ID={track_id} g={growth:.2f} "
            f"v={corrected_speed:.3f} raw={raw_speed:.3f} "
            f"streak={vehicle_streak[key]}"
        )

    if key in vehicle_triggered:
        label += " FIRED"

    status = (
        "TRIGGER_NOW" if message is not None
        else "ALREADY_FIRED" if key in vehicle_triggered
        else "NO_TRIGGER"
    )

    debug = {
        "frame": frame_index,
        "time_s": round(time_s, 3),
        "class": name,
        "class_id": class_id,
        "track_id": track_id,
        "missing_frames": missing_frames,
        "history_len": len(history),
        "growth": growth,
        "growth_threshold": GROWTH_THRESHOLD,
        "raw_center_speed": raw_speed,
        "center_speed": corrected_speed,
        "center_speed_max": CENTER_SPEED_MAX,
        "streak": vehicle_streak[key],
        "required_streak": TRIGGER_STREAK,
        "status": status,
        "reason": " | ".join(reasons),
        **camera_info,
    }
    return message, label, debug
```

```python
from collections import defaultdict, deque
from pathlib import Path
from datetime import datetime
from tqdm.auto import tqdm
import numpy as np
import pandas as pd

model = YOLO("yolo11n-seg.pt")
background_motion = BackgroundMotion()

approach_history = defaultdict(lambda: deque(maxlen=APPROACH_WINDOW))
approach_last_seen = {}
vehicle_streak = defaultdict(int)
vehicle_triggered = set()

color_history = defaultdict(lambda: deque(maxlen=5))
last_stable_color = {}
light_triggered = set()

trigger_logs = []
debug_logs = []
camera_logs = []
active_messages = deque()

DEBUG_PRINT_EVERY = 15
debug_last_print = {}

output_dir = (
    Path("trigger_results")
    / datetime.now().strftime("%Y%m%d_%H%M%S_%f")
)
output_dir.mkdir(parents=True, exist_ok=True)

OUTPUT_PATH = str(output_dir / "result_raw.mp4")
DEBUG_PATH = output_dir / "approach_debug.csv"

cap = cv2.VideoCapture(VIDEO_PATH)
if not cap.isOpened():
    raise RuntimeError(f"영상을 열 수 없습니다: {VIDEO_PATH}")

fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

if not np.isfinite(fps) or fps <= 0:
    cap.release()
    raise RuntimeError("영상 FPS를 읽을 수 없습니다.")

writer = None
frame_index = 0
progress = tqdm(total=total_frames, desc="영상 분석")


def draw_text(image, text, position, color=(255, 255, 255)):
    cv2.putText(
        image, text, position,
        cv2.FONT_HERSHEY_SIMPLEX, 0.5,
        (0, 0, 0), 3, cv2.LINE_AA,
    )
    cv2.putText(
        image, text, position,
        cv2.FONT_HERSHEY_SIMPLEX, 0.5,
        color, 1, cv2.LINE_AA,
    )


def format_value(value):
    return "--" if value is None else f"{value:.3f}"


print(
    f"접근 설정: window={APPROACH_WINDOW}, avg={APPROACH_AVG}, "
    f"g>={GROWTH_THRESHOLD}, corrected_v<={CENTER_SPEED_MAX}, "
    f"streak>={TRIGGER_STREAK}"
)

try:
    while True:
        success, frame = cap.read()
        if not success:
            break

        frame_index += 1
        video_time = (frame_index - 1) / fps
        h, w = frame.shape[:2]

        result = model.track(
            frame,
            persist=True,
            tracker="bytetrack.yaml",
            classes=TARGET_CLASSES,
            conf=CONF_THRESHOLD,
            imgsz=IMAGE_SIZE,
            device=DEVICE,
            verbose=False,
        )[0]

        # ID 유무와 관계없이 반환된 bbox를 배경 마스킹에 사용
        mask_boxes = (
            result.boxes.xyxy.cpu().numpy()
            if result.boxes is not None
            else np.empty((0, 4))
        )

        matrix, camera_info = background_motion.update(frame, mask_boxes)

        # 객체가 검출되지 않은 프레임에도 배경 변환을 누적
        update_approach_coordinates(
            matrix, camera_info, frame_index, fps
        )

        camera_logs.append({
            "frame": frame_index,
            "time_s": round(video_time, 3),
            **camera_info,
        })

        if frame_index == 1 or frame_index % DEBUG_PRINT_EVERY == 0:
            tqdm.write(
                f"CAMERA | {video_time:.2f}s | "
                f"{camera_info['camera_reason']} | "
                f"inliers={camera_info['camera_inliers']}/"
                f"{camera_info['camera_points']} | "
                f"ratio={camera_info['camera_inlier_ratio']:.2f}"
            )

        annotated = result.plot()
        scale = min(1.0, 960 / max(h, w))
        out_w = max(2, int(w * scale) // 2 * 2)
        out_h = max(2, int(h * scale) // 2 * 2)
        view = cv2.resize(annotated, (out_w, out_h))
        panel_height = 175

        if writer is None:
            writer = cv2.VideoWriter(
                OUTPUT_PATH,
                cv2.VideoWriter_fourcc(*"mp4v"),
                fps,
                (out_w, out_h + panel_height),
            )
            if not writer.isOpened():
                raise RuntimeError("결과 영상 저장을 시작하지 못했습니다.")

        if result.boxes is not None and result.boxes.id is not None:
            boxes = result.boxes.xyxy.cpu().numpy()
            ids = result.boxes.id.int().cpu().tolist()
            classes = result.boxes.cls.int().cpu().tolist()

            for box, track_id, class_id in zip(boxes, ids, classes):
                x1, y1, x2, y2 = box
                message = None
                label = ""

                # 1. 차량·장애물 접근
                if (
                    class_id in VEHICLE_CLASSES
                    or class_id in OBSTACLE_CLASSES
                ):
                    message, label, debug = check_approach(
                        box, class_id, track_id,
                        frame_index, fps, w, h, camera_info,
                    )
                    debug_logs.append(debug)

                    should_print = (
                        frame_index - debug_last_print.get(
                            track_id, -DEBUG_PRINT_EVERY
                        ) >= DEBUG_PRINT_EVERY
                        or message is not None
                        or debug["missing_frames"] > 0
                    )

                    if should_print:
                        debug_last_print[track_id] = frame_index
                        tqdm.write(
                            f"DEBUG | {video_time:.2f}s | "
                            f"{debug['class']} ID={track_id} | "
                            f"g={format_value(debug['growth'])} | "
                            f"v_raw={format_value(debug['raw_center_speed'])} | "
                            f"v={format_value(debug['center_speed'])}/"
                            f"{CENTER_SPEED_MAX:.3f} | "
                            f"streak={debug['streak']}/{TRIGGER_STREAK} | "
                            f"{debug['status']} | {debug['reason']}"
                        )

                # 2. 신호등 색상 변화
                elif class_id == TRAFFIC_LIGHT_CLASS:
                    key = (class_id, track_id)

                    bx1 = max(0, min(int(x1), w))
                    bx2 = max(0, min(int(x2), w))
                    by1 = max(0, min(int(y1), h))
                    by2 = max(0, min(int(y2), h))

                    current_color = (
                        "unknown"
                        if bx2 <= bx1 or by2 <= by1
                        else classify_light_color(frame, box)
                    )

                    history = color_history[key]
                    history.append(current_color)

                    stable_color = None
                    if len(history) >= 3:
                        recent = list(history)[-3:]
                        if (
                            recent[0] == recent[1] == recent[2]
                            and recent[0] in ("red", "green")
                        ):
                            stable_color = recent[0]

                    previous_color = last_stable_color.get(key)

                    if stable_color is not None:
                        event_key = (
                            class_id, track_id,
                            previous_color, stable_color,
                        )
                        if (
                            previous_color in ("red", "green")
                            and stable_color != previous_color
                            and event_key not in light_triggered
                        ):
                            light_triggered.add(event_key)
                            message = (
                                f"light ID={track_id} "
                                f"{previous_color}->{stable_color}"
                            )

                        last_stable_color[key] = stable_color

                    label = (
                        f"ID={track_id} raw={current_color} "
                        f"stable={last_stable_color.get(key, '--')}"
                    )

                text_x = max(0, min(out_w - 20, int(x1 * out_w / w)))
                text_y = max(
                    20, min(out_h - 10, int(y2 * out_h / h) + 20)
                )
                draw_text(view, label, (text_x, text_y), (0, 255, 255))

                if message is not None:
                    trigger_logs.append({
                        "frame": frame_index,
                        "time_s": round(video_time, 3),
                        "class_id": class_id,
                        "track_id": track_id,
                        "message": message,
                    })
                    active_messages.append((video_time, message))
                    tqdm.write(
                        f"TRIGGER | {video_time:.2f}s | "
                        f"frame={frame_index} | {message}"
                    )

        while active_messages and video_time - active_messages[0][0] > 2:
            active_messages.popleft()

        panel = np.zeros((panel_height, out_w, 3), dtype=np.uint8)

        draw_text(
            panel,
            f"Time={video_time:.2f}s Frame={frame_index} "
            f"Events={len(trigger_logs)}",
            (10, 25),
        )
        draw_text(
            panel,
            f"Camera={camera_info['camera_reason']} "
            f"Inliers={camera_info['camera_inliers']}/"
            f"{camera_info['camera_points']}",
            (10, 50),
            (0, 255, 0) if camera_info["camera_ok"] else (0, 165, 255),
        )
        draw_text(
            panel,
            "TRIGGER" if active_messages else "NO NEW TRIGGER",
            (10, 75),
            (0, 0, 255) if active_messages else (180, 180, 180),
        )

        for i, (event_time, event_message) in enumerate(
            list(active_messages)[-3:]
        ):
            draw_text(
                panel,
                f"{event_time:.2f}s {event_message}",
                (10, 100 + i * 25),
                (0, 0, 255),
            )

        writer.write(np.vstack([panel, view]))
        progress.update(1)

finally:
    cap.release()
    if writer is not None:
        writer.release()
    progress.close()

    log_df = pd.DataFrame(
        trigger_logs,
        columns=["frame", "time_s", "class_id", "track_id", "message"],
    )
    debug_df = pd.DataFrame(debug_logs)
    camera_df = pd.DataFrame(
        camera_logs,
        columns=[
            "frame", "time_s", "camera_ok", "camera_reason",
            "camera_points", "camera_inliers", "camera_inlier_ratio",
        ],
    )

    log_df.to_csv(
        output_dir / "triggers.csv",
        index=False, encoding="utf-8-sig",
    )
    debug_df.to_csv(
        DEBUG_PATH,
        index=False, encoding="utf-8-sig",
    )
    camera_df.to_csv(
        output_dir / "camera_debug.csv",
        index=False, encoding="utf-8-sig",
    )

print(f"\n완료: {frame_index}프레임 / 트리거 {len(trigger_logs)}회")
print("결과 영상:", OUTPUT_PATH)
print("접근 진단 로그:", DEBUG_PATH)
print("카메라 진단 로그:", output_dir / "camera_debug.csv")

display(log_df)

if not debug_df.empty:
    display(debug_df.tail(50))

if not camera_df.empty:


    display(camera_df["camera_reason"].value_counts())
```
